# Frequency downsampler
This notebook create several copies of already existing data in CSV created with notebook "1- Dataset preparation" from this same project, each copy being a downsample of the original (frequency divided by 2, 4, 8, 16, 32... to a minimum of 20Hz)

```python
import sys

print("Python version {}.{}".format(sys.version_info.major, sys.version_info.minor))

sys.path.insert(1, './Python_SDK/')

# More imports
import matplotlib.pyplot as plt #for plotting raw data
import os #for system folders and files management
import pathlib #for system folders and files management
import shutil #copying files

from IPython.display import display # for folder selection dialog
from ipyfilechooser import FileChooser # for folder selection dialog

import numpy as np
import math
import csv

import pandas as pd #for Dataframe handling

import re #for character extraction

# Make numpy values easier to read (optional)
np.set_printoptions(precision=3, suppress=True)

# End
```

```python
# Create and display a FileChooser widget
fc1 = FileChooser('./')
fc1.show_only_dirs=True
print("--------------------\nSELECT SOURCE FOLDER\n--------------------")
display(fc1)
print("--------------------")
```

```python
# Create and display a FileChooser widget
fc2 = FileChooser('./')
fc2.show_only_dirs=True
print("-------------------------\nSELECT DESTINATION FOLDER\n-------------------------")
display(fc2)
print("-------------------------")
```

```python
src_path = fc1.selected_path
dest_path = fc2.selected_path

print("Downsampling is starting... please wait...")  
for i in range(len(os.listdir(src_path))):
    sensor_folder_src=os.listdir(src_path)[i] #sensor folder name
     
    freq_src = int(re.sub('[^0-9]','', sensor_folder_src[-7:-2])) #frequency is in the sensor folder name
    
    print("Downsampling sample {}".format(sensor_folder_src))
    
    current_src_folder=os.path.join(src_path,sensor_folder_src) #full sensor folder path(origin, read)
   
    for j in range(len(os.listdir(current_src_folder))): #for every file in sensor folder
        Current_file=os.listdir(current_src_folder)[j]
        Current_file_path = os.path.join(current_src_folder,Current_file) #file path
        
        df = pd.read_csv(Current_file_path, header=None) #load csv as a dataframe
        freq_dst=int(freq_src/2)
        while freq_dst>=10:
            df = df[0::2]
            sensor_folder_dst="{}_to_{}Hz".format(sensor_folder_src,freq_dst)
            sensor_folder_dst_path=os.path.join(dest_path,sensor_folder_dst)
            dst_file=os.path.join(sensor_folder_dst_path,Current_file)
            if not os.path.exists(sensor_folder_dst_path): 
                os.makedirs(sensor_folder_dst_path)            
            df.to_csv(path_or_buf=dst_file,index=False, header=None)
            freq_dst=int(freq_dst/2)
            
        
        
        
print("All samples at {} folder successfully downsampled in folder {}".format(src_path,dest_path))  
```
