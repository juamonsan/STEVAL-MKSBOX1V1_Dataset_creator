# Report merger

This code generates a single CSV report containing all the metrics obtained during the training in notebook 2-Neural_Network_training from this same project.

The models root folder must be selected as the source folder in order for the code to find all .csv results files in each sensor folder.

```python
#----------------------------------------------------------------------
#IMPORTS
#----------------------------------------------------------------------
import sys
import os #for system folders and files management
import pathlib #for system folders and files management
import pandas as pd #for Dataframe handling

from datetime import datetime #getting current time for naming the result file

from IPython.display import display # for folder selection dialog
from ipyfilechooser import FileChooser # for folder selection dialog

import numpy as np
import math
import csv

import re #for character extraction
#----------------------------------------------------------------------
# IMPORTS END
#----------------------------------------------------------------------
```

```python
#----------------------------------------------------------------------
# DEFINITIONS
#----------------------------------------------------------------------





#----------------------------------------------------------------------
# DEFINITIONS END
#----------------------------------------------------------------------
```

```python
# Create and display a FileChooser widget
fc1 = FileChooser('./')
fc1.show_only_dirs=True
print("--------------------\nSELECT SOURCE FOLDER\n--------------------")
display(fc1)
print("--------------------")
```

```python
# Create and display a FileChooser widget
fc2 = FileChooser('./')
fc2.show_only_dirs=True
print("--------------------\nSELECT DESTINATION FOLDER\n--------------------")
display(fc2)
print("--------------------")
```

```python
#----------------------------------------------------------------------
# MAIN CODE START
#----------------------------------------------------------------------

now = datetime.now() # current date and time
# root_folder="DATA\\models\\Primera Prueba"
root_folder=fc1.selected_path
# dest_file="DATA\\Results.csv"
dest_file=os.path.join(fc2.selected_path,"Results_{}.csv".format(now.strftime("%Y%m%d_%H%M%S")))

df = pd.DataFrame(columns=['Sensor Type',
                           'Sensor Name',
                           'Frequency',
                           'RNN Type',
                           'Layers',
                           'Nodes 1st Layer',
                           'Nodes 2nd Layer',
                           'Trainable Param',
                           'Non Trainable Param',
                           'Total Param',
                           'F1 Overall Accuracy',
                           'F1 Macro Average',
                           'F1 Weighted Average'])

index_report=0
index_summary=0

#List creation
Sensor_type=[]
Sensor_name=[]
Frequency=[]
RNN_type=[]
Layers=[]
nodes_1=[]
nodes_2=[]
T_Param=[]
NT_Param=[]
Tot_Param=[]
F1_Overall=[]
F1_Macro=[]
F1_Weighted=[]
               

for i in range(len(os.listdir(root_folder))): #access folders
    current_folder=os.path.join(root_folder,os.listdir(root_folder)[i]) 
    folder_name=os.path.basename(os.path.normpath(current_folder)) #folder name could be used for sensor type and name
    folder_name_listed=folder_name.split('_')
    for j in range(len(os.listdir(current_folder))):
        current_file=os.path.join(current_folder,os.listdir(current_folder)[j])
        current_file_name=os.path.basename(os.path.normpath(current_file))
        
        if current_file_name[:6]=="Report":
            
            Sensor_type.insert(index_report,folder_name_listed[0])
            Sensor_name.insert(index_report,folder_name_listed[1])
            Sens_info=current_file_name[:-6].split('_')
            Layers.insert(index_report,int(Sens_info[2][:-1]))
            nodes_1.insert(index_report,int(Sens_info[3]))
            if Layers[index_report]==1:
                nodes_2.insert(index_report,int(0))
            else:
                nodes_2.insert(index_report,int(Sens_info[4]))
            Frequency.insert(index_report,int(Sens_info[-1]))
            RNN_type.insert(index_report,Sens_info[1])
            
            temp_df=pd.read_csv(current_file)
            
            F1_Overall.insert(index_report,temp_df.iloc[3,3])
            F1_Macro.insert(index_report,temp_df.iloc[4,3])
            F1_Weighted.insert(index_report,temp_df.iloc[5,3])
            
            folder_name
            
            index_report=index_report+1
            


            

            
        
        elif current_file_name[:7]=="Summary":
            
            with open(current_file, 'r') as f:
                T_Param.insert(index_summary, int(re.sub('[^0-9]','', f.readlines()[-3])))
            with open(current_file, 'r') as f:
                NT_Param.insert(index_summary, int(re.sub('[^0-9]','', f.readlines()[-2])))
            f.close()
            Tot_Param.insert(index_summary,T_Param[index_summary]+NT_Param[index_summary])
            index_summary=index_summary+1

            
zipped = list(zip(Sensor_type, Sensor_name, Frequency,
               RNN_type, Layers, nodes_1, nodes_2,
               T_Param,NT_Param,Tot_Param,
               F1_Overall,F1_Macro,F1_Weighted))

df = pd.DataFrame(zipped, columns=['Sensor Type',
                           'Sensor Name',
                           'Frequency',
                           'RNN Type',
                           'Layers',
                           'Nodes 1st Layer',
                           'Nodes 2nd Layer',
                           'Trainable Param',
                           'Non Trainable Param',
                           'Total Param',
                           'F1 Overall Accuracy',
                           'F1 Macro Average',
                           'F1 Weighted Average'])

df.to_csv(dest_file, index=False)

df
```
