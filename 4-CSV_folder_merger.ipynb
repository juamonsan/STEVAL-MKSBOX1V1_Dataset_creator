# CSV Folder merger

This notebook contains code for merging 2 directories created with notebook "1- Dataset preparation" from this same project into 1.

```python
import sys

print("Python version {}.{}".format(sys.version_info.major, sys.version_info.minor))

sys.path.insert(1, './Python_SDK/')

# More imports
import matplotlib.pyplot as plt #for plotting raw data
import os #for system folders and files management
import pathlib #for system folders and files management
import shutil #copying files

from IPython.display import display # for folder selection dialog
from ipyfilechooser import FileChooser # for folder selection dialog

import numpy as np
import math
import csv

# Make numpy values easier to read (optional)
np.set_printoptions(precision=3, suppress=True)

# End
```

```python
# Create and display a FileChooser widget
fc1 = FileChooser('./')
fc1.show_only_dirs=True
print("--------------------\nSELECT SOURCE FOLDER\n--------------------")
display(fc1)
print("--------------------")
```

```python
# Create and display a FileChooser widget
fc2 = FileChooser('./')
fc2.show_only_dirs=True
print("-------------------------\nSELECT DESTINATION FOLDER\n-------------------------")
display(fc2)
print("-------------------------")
```

```python
# TODO: ERROR HANDLING:
# check if folders and samples are the same
# Check if samples have same size...
```

```python
src_path = fc1.selected_path
dest_path = fc2.selected_path

print("Merging folders... please wait...")  
for i in range(len(os.listdir(src_path))):
    current_dst_folder=os.path.join(dest_path,os.listdir(dest_path)[i]) #Sensor folder (destination, write)
    destination_files_number=len(os.listdir(current_dst_folder))#number of files in that folder
    current_src_folder=os.path.join(src_path,os.listdir(src_path)[i]) #sensor folder (origin, read)
    for j in range(len(os.listdir(current_src_folder))): #for every file in sensor folder
        Current_file_path = os.path.join(current_src_folder,os.listdir(current_src_folder)[j]) #file path              
        new_filename="{0:0>4}_{1}.csv".format(int(Current_file_path[-11:-7])+destination_files_number,Current_file_path[-6:-4])  #maintain class, but increment filenumber             
        destination=os.path.join(current_dst_folder,new_filename) #new file in destination folder       
        shutil.copy(Current_file_path, destination) #copy renamed file
        
print("Successfully merged {} folder with {}".format(src_path,dest_path))    
```
